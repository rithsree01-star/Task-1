```python
df["date"] = pd.to_datetime(df["date"], utc=True)

df["date"].head()
```

```python
df = df.sort_values("date")

df.head()
```

```python
df["close"].mean()
```

```python
df["close"].min()
```

```python
df["Daily_Change"] = df["close"].diff()

df[["date", "close", "Daily_Change"]].head()
```

```python
df["volume"].mean()
```

```python
df["MA_20"] = df["close"].rolling(window=20).mean()

df["MA_50"] = df["close"].rolling(window=50).mean()

df[["date", "close", "MA_20", "MA_50"]].tail()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["date"], df["close"])

plt.title("Shopify Closing Stock Price Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["close"])

plt.title("Box Plot of Shopify Closing Prices")
plt.xlabel("Closing Price")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["date"].tail(10).astype(str),
        df["close"].tail(10))

plt.title("Shopify Closing Stock Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["close"], bins=20)

plt.title("Shopify Closing Price Distribution")
plt.xlabel("Closing Price")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["open"], df["close"])

plt.title("Opening Price vs Closing Price")
plt.xlabel("Opening Price")
plt.ylabel("Closing Price")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

selected_columns = df[["open", "high", "low",
                       "close", "volume"]]

sns.heatmap(selected_columns.corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f",
            linewidths=0.5)

plt.title("Shopify Stock Correlation Heatmap")

plt.show()
```
