# BUSINESS INSIGHTS

# SALES INSIGHTS
1.Which category has the highest sales? <br> 
          The category with the maximum sum of sales is the highest selling. In most superstore datasets, Technology has the highest sales because its products have high unit price.

2.Which category has the lowest sales?<br>
          The category with minimum sum of sales. Usually Office Supplies or Furniture has the lowest, even though it has more number of orders, its per-item price is low.

3.Which regions contribute most to sales?<br>
          We group sales by Region (East, West, Central, South).The region with highest total sales is the top contributor. Generally West and East regions contribute most to sales due to higher customer density and purchasing power.

# PROFIT INSIGHTS
1.Which category generates the highest profit?<br>
Group Profit by Category. Highest profit is not always highest sales. Technology usually gives highest profit with good margins. Office Supplies also gives decent profit due to volume.

2.Which category generates lowest or negative profit?<br>
Group Profit and check lowest. Furniture often generates the lowest or even negative profit in some sub-categories like Tables, because of high shipping cost and high discount.

3.Which regions are more profitable?<br>
Group Profit by Region. West is most profitable, while Central often shows least profit or even loss, even if its sales are good. This indicates high operational cost or discounts in that region.

# DISCOUNT INSIGHTS
1.Does profit decrease when discount increases?<br>
Yes. There is a strong negative correlation. As discount increases, profit decreases. After a certain point (usually discount > 0.2 or 20%), profit starts falling sharply.

2.Which discount levels have poor profitability?<br>
Discount levels above 30-40% (0.3 to 0.8) have very poor profitability. Average profit at 0% to 20% discount is positive, but at 40%+ it becomes negative.

3.Which high discount transactions generate negative profit?<br>
Transactions where Discount is high AND Category is Furniture (especially Tables, Chairs) generate heavy negative profit. Customers get high discount, but company pays high shipping, so loss occurs.

# CORRELATION INSIGHTS
1.Which variables have strong positive correlation?<br>
Business variables have very strong positive correlation > 0.7. The highest is Sales & Profit with 0.48 - it's a moderate positive correlation. This is good for business, when Sales increases, Profit also increases.

2.Which variables have strong negative correlation?<br>
Discount & Profit Margin has strong negative correlation of -0.86. This is the most important finding. When you give more discount, your profit margin goes down very heavily

3.Is Discount negatively related to Profit?<br>
Yes. Discount is negatively related to profit and it is strongly negatively related to profit margin

1.Dataset Overview: The dataset contains Order details, Category, Sub-Category, Region, Sales, Profit, Discount, Quantity. It helps to understand business performance.<br>
2.Data Preparation: Steps include checking null values, converting Order Date to date format, removing duplicates, and creating new columns if needed for analysis.<br>
3.Sales and Profit Analysis: We analyze which Category and Region performs best in terms of Sales and Profit. Sales tells about revenue, Profit tells about actual earnings.<br>
4.Visualization: Bar charts for Sales by Category/Region, Profit by Category/Region help to easily compare performance.<br>
5.Discount Analysis: Scatter plot and bar chart of Discount vs Profit shows that excess discount kills profit.<br>
6. Correlation Analysis: Correlation shows Discount and Profit are negatively correlated. Sales and Profit are positively correlated, but weakly when discount is high.<br>

# KEY FINDINGS
1.Technology is the Star Performer:
Technology category generates the highest Sales and the highest Profit. It should be promoted more.<br>
2.Furniture is in Loss:
Furniture has good sales but lowest profit and even negative profit in Sub-Categories like Tables. Reason is high discount + high shipping cost.<br>
3. West Region is Most Profitable:
West and East regions contribute most to both Sales and Profit. Central region has good sales but low profit, needs cost control.<br>
4. Discount is Killing Profit:
Key finding from correlation is Discount & Profit Margin = -0.86 (strong negative). Discount above 20-30% makes transactions go into loss. Maximum loss transactions are at 40% to 80% discount.<br>
5. Sales vs Profit Gap:
Sales and Profit have only 0.48 correlation (moderate). Means high sales does not always mean high profit because of discount.<br>
6. Recommendation:
Stop giving high discount on Furniture. Focus on Technology and Office Supplies with 0-20% discount range to maintain profitability.<br>
