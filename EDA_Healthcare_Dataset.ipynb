```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Convert dates and create Length of Stay (needed for heatmap)
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
df['Length of Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

# 2. Select only numeric columns for heatmap
numeric_df = df[['Age', 'Billing Amount', 'Room Number', 'Length of Stay']]

# 3. Plot heatmap
plt.figure(figsize=(7,5))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap - Healthcare Dataset')
plt.tight_layout()
plt.show()
```
