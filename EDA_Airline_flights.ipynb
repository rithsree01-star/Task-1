# TASK 2: UNIVARIATE ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (8, 5)

# Assuming df is already loaded from Task 1
# df = pd.read_csv('airlines_flights_data.csv')

# 1. AIRLINE - Categorical → Count Plot
plt.figure()
sns.countplot(data=df, x='airline', order=df['airline'].value_counts().index)
plt.title('Distribution of Flights by Airline')
plt.xticks(rotation=45)
plt.ylabel('Number of Flights')
plt.show()

# 2. CLASS - Categorical → Count Plot
plt.figure()
sns.countplot(data=df, x='class', order=df['class'].value_counts().index)
plt.title('Distribution of Flights by Class')
plt.ylabel('Number of Flights')
plt.show()

# 3. STOPS - Categorical → Count Plot
plt.figure()
sns.countplot(data=df, x='stops', order=df['stops'].value_counts().index)
plt.title('Distribution of Flights by Number of Stops')
plt.ylabel('Number of Flights')
plt.show()

# 4. DURATION - Numeric → Histogram + Boxplot
fig, axes = plt.subplots(1, 2)
sns.histplot(data=df, x='duration', kde=True, ax=axes[0])
axes[0].set_title('Histogram of Duration')

sns.boxplot(data=df, x='duration', ax=axes[1])
axes[1].set_title('Boxplot of Duration')
plt.tight_layout()
plt.show()

# 5. PRICE - Numeric → Histogram + Boxplot
fig, axes = plt.subplots(1, 2)
sns.histplot(data=df, x='price', kde=True, ax=axes[0])
axes[0].set_title('Histogram of Price')

sns.boxplot(data=df, x='price', ax=axes[1])
axes[1].set_title('Boxplot of Price')
plt.tight_layout()
plt.show()
```

# TASK 3: BIVARIATE ANALYSIS

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# df = pd.read_csv('airlines_flights_data.csv') # Make sure df is already loaded

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 1. PRICE vs AIRLINE - Boxplot
plt.figure()
sns.boxplot(data= df, x='airline', y='price')
plt.title('Price Distribution by Airline')
plt.xticks(rotation=45)
plt.ylabel('Price')
plt.show()

# 2. PRICE vs CLASS - Boxplot + Violin
plt.figure()
sns.boxplot(data= df, x='class', y='price')
plt.title('Price Distribution by Travel Class')
plt.ylabel('Price')
plt.show()

# 3. PRICE vs STOPS - Boxplot
plt.figure()
sns.boxplot(data=df, x='stops', y='price')
plt.title('Price Distribution by Number of Stops')
plt.xlabel('Number of Stops')
plt.ylabel('Price')
plt.show()

# 4. PRICE vs DURATION - Scatter Plot
plt.figure()
sns.scatterplot(data=df, x='duration', y='price', alpha=0.5)
plt.title('Price vs Flight Duration')
plt.xlabel('Duration')
plt.ylabel('Price')
plt.show()

# 5. PRICE vs DAYS_LEFT - Scatter Plot + Regression line
plt.figure()
sns.regplot(data=df, x='days_left', y='price', scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Price vs Days Left Before Flight')
plt.xlabel('Days Left')
plt.ylabel('Price')
plt.show()

# Bonus: Correlation Heatmap for numeric columns
plt.figure(figsize=(8,6))
sns.heatmap(df[['price', 'duration', 'days_left']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

# TASK 4: CORRELATION AND FEATURE ENGINEERING

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Check columns
print("Columns:", df.columns.tolist())

# 2. Replace 'Price' with your actual column name
price_col = 'price' # <-- change this to match your column

# 3. Encode categoricals
df_encoded = df.copy()
for col in df.select_dtypes(include=['object']).columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

# 4. Correlation heatmap - only numeric columns
plt.figure(figsize=(12,8))
corr = df_encoded.corr()
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()

# 5. Features most related to price
if price_col in corr.columns:
    price_corr = corr[price_col].sort_values(key=abs, ascending=False)
    print("Top features correlated with", price_col, ":\n", price_corr.head(10))
else:
    print(f"Column '{price_col}' not found. Check column names above")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Auto-detect price column
price_keywords = ['price', 'fare', 'cost', 'ticket']
price_col = [col for col in df.columns if any(k in col.lower() for k in price_keywords)][0]
print("Using column:", price_col)

# Auto-detect other common columns
airline_col = [col for col in df.columns if 'airline' in col.lower()][0]
source_col = [col for col in df.columns if 'source' in col.lower() or 'from' in col.lower()][0]
dest_col = [col for col in df.columns if 'dest' in col.lower() or 'to' in col.lower()][0]

# 1. EDA
print(df.info())
print("\nMissing Values:\n", df.isnull().sum())

# 2. PLOTS
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Insight 1: Price distribution
sns.histplot(df[price_col], bins=30, kde=True, ax=axes[0,0])
axes[0,0].set_title(f'Distribution of {price_col}')

# Insight 2: Avg price by Airline
df.groupby(airline_col)[price_col].mean().sort_values().plot(kind='barh', ax=axes[0,1])
axes[0,1].set_title(f'Avg {price_col} by Airline')

plt.tight_layout()
plt.show()
```
