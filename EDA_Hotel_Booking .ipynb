```python
df["hotel"].value_counts().plot(kind="bar")

plt.title("Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
df["is_canceled"].value_counts().plot(kind="bar")

plt.title("Hotel Booking Cancellations")
plt.xlabel("Cancellation Status (0 = No, 1 = Yes)")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
df["lead_time"].plot(kind="hist", bins=30)

plt.title("Distribution of Booking Lead Time")
plt.xlabel("Days Between Booking and Arrival")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.countplot(data=df, x="hotel", hue="is_canceled")

plt.title("Cancellations by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

selected_columns = df[["lead_time", "adr", "adults",
                       "children", "babies", "is_canceled"]]

sns.heatmap(selected_columns.corr(), annot=True, cmap="coolwarm",
            fmt=".2f", linewidths=0.5)

plt.title("Hotel Booking Correlation Heatmap")

plt.show()
```
