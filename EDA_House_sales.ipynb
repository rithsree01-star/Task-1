```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='bedrooms',
    y='price',
    estimator='mean',
    errorbar=None,
    hue='bedrooms',
    palette='viridis',
    legend=False
)

plt.title("Average House Price by Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average House Price")
plt.tight_layout()
plt.show()
```

```python
year_price = df.groupby('yr_built')['price'].mean().reset_index()

plt.figure(figsize=(10,5))

sns.lineplot(
    data=year_price,
    x='yr_built',
    y='price',
    marker='o',
    color='blue'
)

plt.title("House Price Trend by Year Built")
plt.xlabel("Year Built")
plt.ylabel("Average House Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='bedrooms', data=df)
plt.title('Number of Bedrooms')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['price'])
plt.title('Price Outliers')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='floors', data=df)
plt.title('Distribution of Floors')
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(
    df[['price', 'bedrooms', 'bathrooms', 'sqft_living']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("House Sales Correlation Heatmap")
plt.show()
```

# BUSINESS INSIGHTS

1. What factors influence house selling prices?

House prices are influenced by factors such as living area, number of bedrooms, bathrooms, location, house condition, and construction quality. Understanding these factors helps businesses determine property values.

2. Which location is best for real estate investment?

Locations with higher average house prices may indicate premium property markets. Businesses can analyze location-based pricing to identify potential investment opportunities.

3. Do larger houses have higher selling prices?

Generally, houses with larger living areas tend to have higher selling prices because they provide more space and comfort to buyers.

4. Are waterfront houses more expensive than normal houses?

Waterfront houses often have higher market value because of their attractive location, views, and limited availability.

5. How does house condition affect selling prices?

Houses in better condition generally attract higher prices because buyers prefer well-maintained properties that require less renovation.

6. Why is house sales analysis important for businesses?

House sales analysis helps real estate companies understand market trends, set competitive prices, identify profitable investment opportunities, and make better business decisions.
