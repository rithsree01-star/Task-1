```python
# BAR PLOT
plt.figure(figsize=(12,6))

brand_price = df.groupby('make')['price'].mean().sort_values()

brand_price.plot(kind='bar', color='skyblue')

plt.title("Average Automobile Price by Brand")
plt.xlabel("Car Brand")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# LINE PLOT
plt.figure(figsize=(8,5))

sns.lineplot(
    data=df.sort_values('horsepower'),
    x='horsepower',
    y='price',
    color='blue'
)

plt.title("Horsepower vs Automobile Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='engine-size',
    y='price',
    hue='fuel-type',
    alpha=0.7
)

plt.title("Engine Size vs Automobile Price")
plt.xlabel("Engine Size")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=df['price'],
    color='lightgreen'
)

plt.title("Automobile Price Outlier Detection")
plt.xlabel("Price")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(
    df[['price', 'horsepower', 'engine-size', 'curb-weight']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Automobile Correlation Heatmap")
plt.show()
```
