```python
# PRACTICAL 2: Python program to use Conditional Statements and Loops

print("=== PRACTICAL 2: Conditionals and Loops ===")

# 2. Check if number is positive using if
num1 = int(input("\nEnter a number to check if positive: "))
if num1 > 0:
    print(f"{num1} is Positive")

# 3. Check even or odd using if-else
num2 = int(input("\nEnter another number to check even/odd: "))
if num2 % 2 == 0:
    print(f"{num2} is Even")
else:
    print(f"{num2} is Odd")

# 4. Assign grade using if-elif-else
marks = int(input("\nEnter student marks out of 100: "))
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"
print(f"Grade: {grade}")

# 5. Verify login using nested if-else
print("\n--- Login ---")
username = input("Enter username: ")
password = input("Enter password: ")

correct_user = "admin"
correct_pass = "1234"

if username == correct_user:
    if password == correct_pass:
        print("Login Successful!")
    else:
        print("Incorrect Password")
else:
    print("Incorrect Username")

# 6. Voting eligibility using one-line if-else
age = int(input("\nEnter your age: "))
result = "Eligible to Vote" if age >= 18 else "Not Eligible to Vote"
print(result)

# 7. Display 1 to 5 using for loop
print("\nNumbers 1 to 5 using FOR loop:")
for i in range(1, 6):
    print(i, end=" ")

# 8. Display 1 to 5 using while loop
print("\n\nNumbers 1 to 5 using WHILE loop:")
j = 1
while j <= 5:
    print(j, end=" ")
    j += 1

# 9. Demonstrate break statement
print("\n\nBreak statement: Loop stops when value becomes 5")
for k in range(1, 11):
    if k == 5:
        break
    print(k, end=" ")

# 10. Demonstrate continue statement
print("\n\nContinue statement: Skip value 5")
for k in range(1, 11):
    if k == 5:
        continue
    print(k, end=" ")

# 11. Demonstrate pass statement
print("\n\nPass statement: Placeholder in loop")
for k in range(1, 6):
    if k == 3:
        pass  # do nothing, just a placeholder
    else:
        print(k, end=" ")

# 12. Display day name using match-case
print("\n\nEnter day number 1-7")
day = int(input("Enter day number: "))
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day number")

print("\n=== Program Ended ===")
```

# PRACTICAL 3

```python
# PRACTICAL 3: Python program to define and call user-defined functions

#  Function without arguments and without return value
def greet():
    print("Hello! This is a function without arguments and return value.")

#  Function with arguments to perform addition
def add(a, b):
    print(f"Sum of {a} and {b} = {a + b}")

#  Function without arguments that returns a value
def get_message():
    return "This function returns a value."

#  Function with arguments that returns product
def multiply(x, y):
    return x * y

#  Function with default argument
def introduce(name, course="Python"):
    print(f"My name is {name} and I am learning {course}.")

#  Function with keyword arguments
def student_details(name, roll, branch):
    print(f"Name: {name}, Roll No: {roll}, Branch: {branch}")

#  Function with variable-length arguments *args
def total_marks(*marks):
    total = sum(marks)
    print(f"Marks: {marks}")
    print(f"Total Marks = {total}")

#  Lambda function to calculate square
square = lambda n: n * n

#  Recursive function to calculate factorial
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

#  Nested function
def outer():
    print("This is the outer function.")
    
    def inner():
        print("This is the inner function called from outer.")
    
    inner()  # calling inner function


#  Execute all the defined functions
if __name__ == "__main__":
    print("=== PRACTICAL 3 OUTPUT ===\n")
    
    greet()  # 2
    
    add(10, 20)  # 3
    
    msg = get_message()  # 4
    print(msg)
    
    prod = multiply(5, 6)  # 5
    print(f"Product of 5 and 6 = {prod}")
    
    introduce("Arun")  # 6 - without passing default value
    introduce("Priya", "Java")  # 6 - with passing value
    
    student_details(roll=101, branch="CSE", name="Ravi")  # 7 - keyword args
    
    total_marks(85, 90, 78, 92)  # 8
    
    num = 7
    print(f"Square of {num} = {square(num)}")  # 9
    
    n = 5
    print(f"Factorial of {n} = {factorial(n)}")  # 10
    
    outer()  # 11
    
    print("\n=== Program Ended ===")  # 13
```

```python
# PRACTICAL 4: Python program to work with lists and list comprehensions

#  Create a list containing integer elements
my_list = [5, 10, 15, 20, 25]
print("Step 2: Original List Created")

#  Display the original list
print("Step 3: Original List:", my_list)

#  Access and display the first and last elements of the list
print("Step 4: First element:", my_list[0])
print("Step 4: Last element:", my_list[-1])

#  Add a new element to the end of the list using append()
my_list.append(30)
print("Step 5: After append(30):", my_list)

#  Insert an element at a specified position using insert()
my_list.insert(2, 12) # insert 12 at index 2
print("Step 6: After insert(2, 12):", my_list)

#  Remove an element from the list using remove()
my_list.remove(15) # removes first occurrence of 15
print("Step 7: After remove(15):", my_list)

#  Find and display the length of the list using len()
print("Step 8: Length of list:", len(my_list))

#  Traverse the list using a for loop and display each element
print("Step 9: Traversing list with for loop:")
for element in my_list:
    print(element, end=" ")
print("\n")

#  Create a new list containing the squares of all elements using list comprehension
squares = [x**2 for x in my_list]
print("Step 10: Squares using list comprehension:", squares)

#  Create and display a list of even numbers using list comprehension
evens = [x for x in my_list if x % 2 == 0]
print("Step 11: Even numbers:", evens)

#  Create and display a list of odd numbers using list comprehension
odds = [x for x in my_list if x % 2!= 0]
print("Step 12: Odd numbers:", odds)

#  Create a new list by doubling each element using list comprehension
doubled = [x*2 for x in my_list]
print("Step 13: Doubled elements:", doubled)

# Create a nested list (matrix) and flatten it into a single list using nested list comprehension
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Step 14: Original Matrix:", matrix)
flattened = [num for row in matrix for num in row]
print("Step 14: Flattened list:", flattened)

#  Display all generated lists
print("\n--- Final Summary of All Lists ---")
print("Modified Original List:", my_list)
print("Squares List:", squares)
print("Even List:", evens)
print("Odd List:", odds)
print("Doubled List:", doubled)
print("Flattened Matrix:", flattened)

print("\nProgram Ended.")
```

```python
# PRACTICAL 12 - Handle missing data using pandas

#Import libraries
import pandas as pd
import numpy as np

# Create a DataFrame containing missing values
data = {
    'Name': ['Amit', 'Priya', None, 'John', 'Sneha'],
    'Age': [20, None, 22, 21, None],
    'Marks': [85, 90, None, 88, 95],
    'City': ['Chennai', 'Mumbai', 'Delhi', None, 'Kolkata']
}
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame:")
print(df)

# Check and display missing values using isnull()
print("\nMissing values (True = Missing):")
print(df.isnull())

# Count missing values in each column
print("\nNumber of missing values per column:")
print(df.isnull().sum())

# Fill all missing values with a default value
print("\nAfter filling all missing with 'Unknown' / 0:")
df_fill_default = df.fillna({'Name':'Unknown', 'Age':0, 'Marks':0, 'City':'Unknown'})
print(df_fill_default)

# Fill missing values in Marks column with column mean
mean_marks = df['Marks'].mean()
df['Marks'] = df['Marks'].fillna(mean_marks)
print(f"\nAfter filling Marks with mean ({mean_marks}):")
print(df)

# Forward fill
print("\nForward Fill (ffill):")
print(df.ffill())

# Backward fill
print("\nBackward Fill (bfill):")
print(df.bfill())

# Remove rows containing missing values using dropna()
print("\nAfter dropping rows with missing values (dropna):")
print(df.dropna())

# Replace a specified value using replace()
print("\nAfter replacing 'Chennai' with 'CBE' using replace():")
df_replaced = df.replace('Chennai', 'CBE')
print(df_replaced)

# Display DataFrame information using info()
print("\nDataFrame Info:")
df.info()

# End
print("\nProgram Ended")
```

```python
# PRACTICAL 13 Write a Python program to group and aggregate data in pandas

# Import Pandas library
import pandas as pd

# Create a DataFrame containing employee details
data = {
    "Department": ["HR", "IT", "HR", "IT", "Sales", "Sales"],
    "Employee": ["Amit", "Rahul", "Priya", "Ankit", "Sakshi", "Riya"],
    "Salary": [30000, 50000, 35000, 60000, 45000, 40000],
    "Experience": [2, 5, 3, 6, 4, 2]
}
df = pd.DataFrame(data)

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Group the data based on Department column using groupby()
grouped = df.groupby("Department")

# Step 6: Calculate and display the total salary for each department
print("\nTotal salary for each department:")
print(grouped["Salary"].sum())

# average salary for each department
print("\nAverage salary for each department:")
print(grouped["Salary"].mean())

# maximum and minimum salary in each department
print("\nMaximum salary in each department:")
print(grouped["Salary"].max())
print("\nMinimum salary in each department:")
print(grouped["Salary"].min())

# Count and display the number of employees in each department
print("\nNumber of employees in each department:")
print(grouped["Employee"].count())

# Perform multiple aggregation operations (sum, mean, max, and min)
print("\nMultiple aggregation operations on salary column:")
print(grouped["Salary"].agg(['sum', 'mean', 'max', 'min']))

# Group the data using both Department and Experience columns
print("\nGroup by Department and Experience and display aggregated salary:")
print(df.groupby(["Department", "Experience"])["Salary"].sum())

# Iterate through each group and display the grouped records
print("\nIterate through each group:")
for name, group in grouped:
    print(f"\nDepartment: {name}")
    print(group)

# Sort the DataFrame by the Salary column and display the sorted data
print("\nDataFrame sorted by Salary:")
print(df.sort_values(by="Salary"))

# End the program
print("\nProgram Ended")
```

```python
# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\\Users\\riths\\OneDrive\\Desktop\\Weekly Task\students.csv"


# Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

#  Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# Reset the index
df = df.reset_index(drop=True)

# Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\\Users\\riths\\OneDrive\\Desktop\\Weekly Task\students.csv"


# Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# Display the dataset
print("========== DATASET ==========")
print(df)

# Select Marks column
marks = df["Marks"]

# Remove missing values
marks = marks.dropna()

# Count number of observations
count = marks.count()

# Calculate total marks
total = marks.sum()

# Calculate Mean
mean = marks.mean()

# Calculate Median
median = marks.median()

# Calculate Mode
mode = marks.mode()

# Calculate Standard Deviation
std = marks.std()

# Calculate Variance
variance = marks.var()

# Find Minimum
minimum = marks.min()

# Find Maximum
maximum = marks.max()

# Calculate Range
data_range = maximum - minimum

# Calculate First Quartile
q1 = marks.quantile(0.25)

# Calculate Third Quartile
q3 = marks.quantile(0.75)

# Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```
